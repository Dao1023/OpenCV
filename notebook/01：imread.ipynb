```python
import cv2

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

img = cv2.imread('../assets/images/1.jpg', 1)
# cv2.imshow('test',img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title(f'BGR Image: {img.shape}')
plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title(f'RGB Image: {img_rgb.shape}')
```

```python
ret, thresh1 = cv2.threshold(img_rgb, 127, 255, cv2.THRESH_BINARY)
ret, thresh2 = cv2.threshold(img_rgb, 127, 255, cv2.THRESH_BINARY_INV)
ret, thresh3 = cv2.threshold(img_rgb, 127, 255, cv2.THRESH_TRUNC)
ret, thresh4 = cv2.threshold(img_rgb, 127, 255, cv2.THRESH_TOZERO)
ret, thresh5 = cv2.threshold(img_rgb, 127, 255, cv2.THRESH_TOZERO_INV)

titles = ['Original Image', 'BINARY', 'BINARY_INV', 'TRUNC', 'TOZERO', 'TOZERO_INV']
images = [img_rgb, thresh1, thresh2, thresh3, thresh4, thresh5]

plt.figure(figsize=(10, 10))
for i in range(6):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], 'gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])
```
